## hypothesis

- H₀ (Null Hypothesis):
There is no significant relationship between maternal health indicators (Age, SystolicBP, DiastolicBP, BS, BodyTemp, HeartRate) and the RiskLevel of maternal mortality.

- H₁ (Alternative Hypothesis):
At least one of these factors has a significant relationship with RiskLevel.

```python
maternal_df['RiskLevel'].unique()
```

```python
features = ['Age','SystolicBP','DiastolicBP','BS','BodyTemp','HeartRate']

for col in features:
    low = maternal_df[maternal_df['RiskLevel']=='low risk'][col]
    mid = maternal_df[maternal_df['RiskLevel']=='mid risk'][col]
    high = maternal_df[maternal_df['RiskLevel']=='high risk'][col]
    print(high)

    f_stat, p_value = stats.f_oneway(low, mid, high)

    print(f"{col}: F={f_stat:.4f}, p={p_value:.4f}")
```

insight:
- p-value < 0.05 for all features hence Reject the null hypothesis (H₀) for every feature
- F-statistic measures the difference between risk groups relative to variation within groups:
- Age: F = 46.114 which is  very high hence strong differences between RiskLevel groups
- HeartRate: F = 20.4527which is moderate ,still significant but less than Age

```python
##  SystolicBP vs RiskLevel
low = maternal_df[maternal_df['RiskLevel']=='low risk']['SystolicBP']
mid = maternal_df[maternal_df['RiskLevel']=='mid risk']['SystolicBP']
high = maternal_df[maternal_df['RiskLevel']=='high risk']['SystolicBP']

f_stat, p_value = stats.f_oneway(low, mid, high)

print(f"F-statistic: {f_stat}, p-value: {p_value}")
```

insight:
- The F-statistic (95.9) is very large.The variation between RiskLevel groups is much greater than the variation within low, mid, high risks groups
- The groups (low, mid, high risk) have very different averages for the variable being tested

- p values is extremely small- 0.0000000000000000000000000000000000000729

- p-value < 0.05 so we Reject the Null Hypothesis (H₀)
- There is strong statistical evidence that at least one maternal health indicator (Age, Blood Pressure, Blood Sugar, Body Temperature, or Heart Rate) has a significant relationship with RiskLevel== not reject H1


in conclusion:
Reject the null hypothesis because There is a statistically significant relationship between the maternal health indicators and the RiskLevel of maternal mortality.
